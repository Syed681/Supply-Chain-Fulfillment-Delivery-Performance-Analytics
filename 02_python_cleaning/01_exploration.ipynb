```python
import pandas as pd
```

```python
# 1. Load the raw dataset
df = pd.read_csv(
    "../01_raw_data/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)
```

```python
# 2. Check dataset size

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# 3. View first 5 rows
print("\nFirst 5 rows:")
print(df.head())
```

```python
# 4. Check column names
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# 5. Check data types
print("\nData types:")
print(df.dtypes)
```

```python
# 6. Check missing values
missing_values = df.isnull().sum()

print("\nMissing values:")
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python
# 7. Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# 8. Check columns with only one unique value
unique_counts = df.nunique()
print("\nColumns with only one unique value:")
print(unique_counts[unique_counts == 1])
```

```python
# 9. Check important identifiers
print("\nUnique identifiers:")

print("Order Item Id:", df["Order Item Id"].nunique())
print("Order Id:", df["Order Id"].nunique())
print("Customer Id:", df["Customer Id"].nunique())
print("Product Card Id:", df["Product Card Id"].nunique())
```

```python
# 10. Check date range
print("\nOrder date range:")
print("Minimum:", df["order date (DateOrders)"].min())
print("Maximum:", df["order date (DateOrders)"].max())

print("\nShipping date range:")
print("Minimum:", df["shipping date (DateOrders)"].min())
print("Maximum:", df["shipping date (DateOrders)"].max())
```

```python
# 11. Check potentially duplicate columns

print("\nChecking potentially duplicate columns:")

print(
    "Customer Id == Order Customer Id:",
    (df["Customer Id"] == df["Order Customer Id"]).all()
)

print(
    "Category Id == Product Category Id:",
    (df["Category Id"] == df["Product Category Id"]).all()
)

print(
    "Product Card Id == Order Item Cardprod Id:",
    (df["Product Card Id"] == df["Order Item Cardprod Id"]).all()
)

print(
    "Sales per customer == Order Item Total:",
    (df["Sales per customer"] == df["Order Item Total"]).all()
)

print(
    "Benefit per order == Order Profit Per Order:",
    (df["Benefit per order"] == df["Order Profit Per Order"]).all()
)

print(
    "Product Price == Order Item Product Price:",
    (df["Product Price"] == df["Order Item Product Price"]).all()
)
```
